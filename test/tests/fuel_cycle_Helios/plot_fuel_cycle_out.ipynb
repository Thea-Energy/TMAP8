```python
import matplotlib.pyplot as plt
import numpy as np
import csv
```

```python
#filename = 'outputs/fuel_cycle_orig_abdou.csv'
filename = 'outputs/fuel_cycle_helios_base_final.csv'

with open(filename, mode ='r')as file:
    csvFile = csv.reader(file)
    i = 0
    data = []
    for line in csvFile:
        if i == 0:
            headers = np.array(line)
        elif i == 1:
            data = [np.array(line, dtype=np.float32)]
        else:
            data.append(np.array(line, dtype=np.float32))
        i = i + 1
    final = np.stack(data, axis=1)
```

```python
#print(type(final))
#print(final[0])
#print(final[:,0])
print(headers)
```

```python
# Recreate Abdou log plot
plt.yscale('log')
plt.xscale('log')
plt.xlim((1e-1,1e4))
plt.ylim((1e-3,1e4))
time_days = final[0] / 86400
#plt.plot(time_days, final[2], label='total tritium')
plt.plot(time_days, final[3], label='T_01_BZ')
plt.plot(time_days, final[11], label='T_09_ISS')
plt.plot(time_days, final[13], label='T_11_storage')
plt.plot(time_days, final[4], label='T_02_TES')
plt.legend()
plt.xlabel('Time (days)')
plt.ylabel('Tritium Inventory (kg)')
plt.grid(True)
plt.savefig('tmap_baseline.pdf')
```
